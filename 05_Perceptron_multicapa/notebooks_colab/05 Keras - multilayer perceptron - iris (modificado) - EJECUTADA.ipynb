```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
target = iris.target
Y = to_categorical(target).astype("uint8")
# 0 => [1 0 0]
# 1 => [0 1 0]
# 2 => [0 0 1]
```

```python
idx = 0
print(X[idx])
print(target[idx])
print(Y[idx])
```

```python
model = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
  ]
)

model.summary()
```

```python
y = model(X)
print(y[149])
```

```python
optimizer=tf.keras.optimizers.SGD(learning_rate=0.03)
loss=tf.keras.losses.MeanSquaredError()

model.compile(optimizer, loss)
history = model.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history['loss'])
plt.show()
```

```python
pd.DataFrame(history.history).plot(figsize=(8,5))
plt.show()
```

```python
model.predict(np.array([[3,3,1,1]]))
```

## Red profunda (4 x 3 x 3 x 3 x 3)

Misma arquitectura de antes mas **dos capas ocultas** extra (2 nuevos `Dense(3, activation="sigmoid")` antes de la salida). `model_deep.summary()` debe mostrar 4 capas `Dense`. Mismo optimizador SGD(lr=0.03), misma perdida MSE, mismas 500 epocas que la version original, para que la comparacion sea justa.

```python
model_deep = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
    layers.Dense(3, activation="sigmoid", name="layer3"),
    layers.Dense(3, activation="sigmoid", name="layer4"),
  ]
)

model_deep.summary()
```

```python
optimizer_deep = tf.keras.optimizers.SGD(learning_rate=0.03)
loss_deep = tf.keras.losses.MeanSquaredError()

model_deep.compile(optimizer_deep, loss_deep)
history_deep = model_deep.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history['loss'], label='original (4x3x3)')
plt.plot(history_deep.history['loss'], label='profunda (4x3x3x3x3)')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()

print('Loss final original:', history.history['loss'][-1])
print('Loss final profunda:', history_deep.history['loss'][-1])
```

```python
model_deep.predict(np.array([[3,3,1,1]]))
```
